# REINFORCE: Policy Gradient Reinforcement Learning

## 1. Introduction

REINFORCE is a **Monte Carlo policy-gradient algorithm** that learns a policy directly from complete episodes of interaction with an environment.

In reinforcement learning, an agent observes a state, selects an action, receives a reward, and transitions to another state. Its objective is to learn behavior that maximizes the expected cumulative reward.

There are two broad approaches to learning:

- **Value-based methods:** Learn how valuable each action is, as in Q-Learning and DQN.
- **Policy-based methods:** Directly learn which actions to select, as in REINFORCE.

REINFORCE belongs to the second category. It uses a neural network to represent a policy and adjusts the network parameters using the rewards collected from sampled episodes.

## 2. What Does the Policy Represent?

A policy, denoted by $\pi$, specifies how an agent selects actions.

In REINFORCE, the policy is parameterized by a neural network with parameters $\theta$:

$$
\pi_\theta(a\mid s)
=
P(A_t=a\mid S_t=s;\theta)
$$

For a given state $s$, the network outputs a probability distribution over the available actions.

For FrozenLake, there are four actions: Left, Down, Right, and Up. The policy might output:

$$
\pi_\theta(\cdot\mid s)
=
[0.10,\ 0.60,\ 0.20,\ 0.10]
$$

This means the agent selects Down with probability $0.60$, Left with probability $0.10$, and so on.

Unlike a greedy value-based policy, REINFORCE samples an action from this distribution during training.

## 3. The Learning Objective

The goal is to maximize the expected return generated by the policy.

Let $\tau$ denote a complete episode (trajectory), containing states, actions, and rewards. Its return is:

$$
G_0=\sum_{t=0}^{T-1}\gamma^t R_{t+1}
$$

where $T$ is the episode length and $\gamma\in[0,1]$ is the discount factor.

We define the objective:

$$
J(\theta)=\mathbb{E}_{\tau\sim\pi_\theta}[G_0]
$$

The notation $\tau\sim\pi_\theta$ means that trajectories are generated by following the policy parameterized by $\theta$.

**Our objective is to find parameters that maximize $J(\theta)$.** REINFORCE does this using gradient ascent:

$$
\theta\leftarrow\theta+\alpha\nabla_\theta J(\theta)
$$

Here, $\alpha$ is the learning rate. The central challenge is calculating the gradient of expected return with respect to the policy's parameters.

## 4. Why Policy Gradients?

In Q-Learning and DQN, the agent estimates action values and selects actions using those estimates.

REINFORCE instead learns the policy directly:

$$
\pi_\theta(a\mid s)
$$

Suppose an action is sampled from the policy and the resulting episode produces a high return. We want to increase the probability of the actions that contributed to that outcome. If the return is poor, we want to decrease their probability relative to better alternatives.

The policy-gradient method provides a mathematical way to perform this update.

Unlike value-based methods, REINFORCE does not require a Q-table or a target Q-value. It learns from complete sampled episodes and their returns.
